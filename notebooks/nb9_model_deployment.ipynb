```python
X_train = full_df[~full_df["TARGET"].isna()]
y_train = X_train.pop("TARGET")
id_train = X_train.pop("SK_ID_CURR")
X_test = full_df[full_df["TARGET"].isna()]
```

```python
import numpy as np
nessesary_columns = {
    "X_cols": X_train.columns,
    "y_col": "TARGET",
    "id_cols": ["SK_ID_CURR", "SK_ID_BUREAU", "SK_ID_PREV"],
    "num_cols": X_train.select_dtypes(include=[np.number]).columns.tolist(),
    "cat_cols": X_train.select_dtypes(include=["category"]).columns.tolist(),
    "bool_cols": X_train.select_dtypes(include=["bool"]).columns.tolist(),
}

print("Number of columns:", len(nessesary_columns["X_cols"]))

joblib.dump(nessesary_columns, "models/deployment_columns.jbl")
```

# Model deployment
We first refit the best performing model

```python
preprocessor = final_model_helper.preprocessor_for_sklearn(
    columns_path="models/deployment_columns.jbl"
).get_preprocessor()
```

```python
from sklearn.linear_model import LogisticRegression
from lightgbm import LGBMClassifier
from sklearn.ensemble import VotingClassifier
from sklearn.pipeline import Pipeline

class_sk_models = [LogisticRegression(max_iter=1000, random_state=3)]
voter = VotingClassifier(
    estimators=[
        ("lgbm_gbdt", LGBMClassifier(boosting_type="gbdt", verbose=-1)),
    ]
    + [
        (
            type(model).__name__,
            Pipeline([("preprocessor", preprocessor), (type(model).__name__, model)]),
        )
        for model in class_sk_models
    ],
    voting="soft",
    n_jobs=-1,
)


display(voter)
```

```python
voter.fit(X_train, y_train)
joblib.dump(voter, "models/deployment_model.jbl")
```

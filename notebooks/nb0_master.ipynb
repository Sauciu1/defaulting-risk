# Master Notebook - A Foreword

This project focusses on the analysis and model building for [Home Credit Default Risk](https://www.kaggle.com/competitions/home-credit-default-risk/data) dataset.

Due to the scope of the project and need for clear and organised presentation, the project was split into multiple notebooks. Consequently, this notebook serves as the kingpin - providing overarching structured, context, and presenting only the most important conclusions from various notebooks.

Although the project was conducted in an exploratory manner, the master notebook is structured to provide a more systematic flow and a coherent story. Secondary information can thus be found within the supporting notebooks. Note that the notebook number prefixes (e.g. nb0_, nb1_, nb2_) reflect the order they were created in rather than a logical flow; a consequence of branching nature of the project.

To further enhance readability and promote reusability, Majority code was moved from notebooks to custom modules in [helpers](..\helpers). Whenever the inner workings of function used are key for analysis, they are expanded on in markdown.

The reader is highly encouraged to read the master notebook before delving into specifics.

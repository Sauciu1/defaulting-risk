# Master Notebook - A Foreword

This project focusses on the analysis and model building for [Home Credit Default Risk](https://www.kaggle.com/competitions/home-credit-default-risk/data) dataset.

Due to the scope of the project and need for clear and organised presentation, the project was split into multiple notebooks. Consequently, this notebook serves as the kingpin - providing overarching structured, context, and presenting only the most important conclusions from various notebooks.

Although the project was conducted in an exploratory manner, the master notebook is structured to provide a more systematic flow and a coherent story. Secondary information can thus be found within the supporting notebooks. Note that the notebook number prefixes (e.g. nb0_, nb1_, nb2_) reflect the order they were created in rather than a logical flow; a consequence of branching nature of the project.

To further enhance readability and promote reusability, Majority code was moved from notebooks to custom modules in [helpers](..\helpers). Whenever the inner workings of function used are key for analysis, they are expanded on in markdown.

The reader is highly encouraged to read the master notebook before delving into specifics.

### Data cleaning and preprocessing
#### [nb1_main_table_cleaning.ipynb](nb1_main_table_cleaning.ipynb)
* Data cleaning of the main applications table. 
* Focus on getting a rough idea for the nature of 120 features stored within the main `applications` table.
* The data type, description, number of unique values, and distribution for each variable are provided to be referenced in further analysis.
* Insights for further data pre-processing procedures are provided.


#### [nb2_supp_tables_cleaning.ipynb](nb2_supp_tables_cleaning.ipynb)
* Data cleaning for the supporting tables.
* Rough investigation for the 7 supplementary tables.
* The data type, description, number of unique values, and distribution for each variable are provided to be referenced in further analysis.
* The need for data aggregation techniques and unsupervised learning is brought to attention.


#### [nb3_basic_preprocessing.ipynb](nb3_basic_prepocessing.ipynb)
* Created preprocessor for handling variable types for both application and supplementary tables.
* Defined target metric as: **ROC_AUC, precision-recall AUC, f1_macro**.
* Confirmed lightGBM baseline of 0.74262 contest score.


#### [nb4_house_parameters.ipynb](nb4_house_parameters.ipynb)
* Mean, Median, and Mode application features are near identical in ordering. Their choice has minimal effect on model performance.
* Inclusion of house location and neighborhood features has marginal effect on model performance.
* Clustering (HDBSCAN) and related feature engineering was deemed unsuccessful in objectively increasing model performance.

#### [nb5_bureau_balance.ipynb](notebooks\nb5_bureau_balance.ipynb)
* Investigated client bureau balance states.
* Modeled the change of balance states as discrete-time discrete-state Markow process.
    * Derived transition matrix and visualised simplified version.
    * Transition matrix provides 0.57 f1_macro performance.
    * Simplified transition matrix simplified as node.
* Create a multi-output state prediction model for 8 classes.
    * 0.63 f1_macro for first predicted state.
    * 0.53 for second predicted state (comparable to transition matrix) 

#### [nb6_bureau.ipynb](notebooks\nb6_bureau.ipynb)
* Aggregated the columns of bureau table for each application.
* 6 Unique clusters (+1 noise) were identified within the data of previous application.
* Integrated bureau_balance analysis into the overall bureau table.
    * Achieved ROC AUC of 0.75578

#### [nb7_previous_applications.ipynb](notebooks\nb7_previous_applications.ipynb)
* Aggregated `previous_application` as predictor for current dataset.
* Merged 3 other tables `POS_CASH_balance`, `installments_payments`, and `credit_card_balance` into the main application table.
    * Note that more nuanced aggregation, directed towards previous applications table, was not used.
* Achieved ROC AUC of 0.76818



#### [nb8_model_tuning.ipynb](notebooks\nb8_model_tuning.ipynb)

        * HistGradientBoostingClassifier only slightly underperformed to LightGBM (CV ROC AUC of 0.7762 vs 0.7784)
